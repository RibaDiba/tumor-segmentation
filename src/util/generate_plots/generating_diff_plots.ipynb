## Correction Error (CE): Quantifying Robustness Degradation

### Metric Definition

The **Correction Error** measures relative performance degradation when models are evaluated on augmented (corrupted) images:

$$CE = \frac{1 - \text{IoU}_{\text{edge}}}{1 - \text{IoU}_{\text{clean}}}$$

Or equivalently, using error rates:

$$CE = \frac{\text{error}_{\text{corrupted}}}{\text{error}_{\text{clean}}}$$

### Interpretation

- **CE > 1**: Corruption degrades performance. Higher values indicate greater vulnerability to augmentation.
- **CE = 1**: No degradation.
- **CE < 1**: Edge case where corruption improves generalization (rare).

Also, a model could still preform better in absolute, but could be seen as less favorable if it degrades more.
